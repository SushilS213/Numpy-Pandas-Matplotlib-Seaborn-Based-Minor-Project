Merging joining and concatinating

```python
import numpy as np
import pandas as pd
```

```python
var1 = pd.DataFrame({"A":[1,2,3,4],"B":[5,6,7,8]},index=["A","B","C","d"])
var2 = pd.DataFrame({"C":[9,10,11],"D":[13,14,15]})
```

```python
pd.concat([var1,var2],axis=1)
```

```python
var1.join(var2,how='left')
```

```python
var1.join(var2,how='right')
```

Group by aggregation

```python
data ={
    'Categeory':['A','B','A','B','A','A','B','A'],
    'Store':['S1','S2','S1','S1','S2','S2','S1','S2'],
    'Quantity':[100,200,150,250,120,180,200,300],
    'Date':pd.date_range('10/04/2026',periods=8)
}
df =pd.DataFrame(data)
df
```

```python
#Group by category and calcualate the sum of sales 
df.groupby('Categeory')['Quantity'].sum()
```

```python
#Group by Store and Calcualate the sum of Quantitydf.
df.groupby("Store")['Quantity'].sum()
```

```python
#Group by multiple Column
df.groupby(['Categeory','Store'])["Quantity"].sum()
```
