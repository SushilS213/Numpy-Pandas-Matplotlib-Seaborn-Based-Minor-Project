```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(1,5,5).astype(int)
y = x**2
plt.plot(x,y,c='r')
```

```python
plt.subplot(2,2,1)
plt.plot(x,y)
plt.subplot(2,2,4)
plt.plot(y,x)
plt.subplot(2,2,3)
plt.plot(x,y)
plt.subplot(2,2,2)
plt.plot(x,y)
```

```python
fig = plt.figure(figsize=(10,5),dpi=100)
axice1 = fig.add_axes([0.2,0.5,0.8,0.9])
axice1.plot(x,y)
axice2 = fig.add_axes([0.4,0.7,0.4,0.4])   # Its called Object_Oriented plotting
axice2.plot(y,x)
```
