```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
arr = np.array([2,3,4,5,6,8,9,10,12,13,14,16,18,19,23,25,27,201])
type(arr)
```

```python
Q1 = np.percentile(arr,25)
Q3 = np.percentile(arr,75)
IQR = Q3 - Q1
IQR
```

```python
upper_F = Q3 + (1.5*IQR)
Lower_F = Q1 - (1.5*IQR)
```

```python
l = []
for i in arr:
    if i>=Lower_F and i<=upper_F:
        l.append(i)
arr2=np.array(l)  
arr2
      
```

```python
plt.subplot(1,2,1)
sns.boxplot(arr)  # Here Poin is the Outlier
plt.subplot(1,2,2)
sns.boxplot(arr2)
```
