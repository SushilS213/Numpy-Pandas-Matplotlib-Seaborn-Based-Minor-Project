2.Two Top Scorers

```python
high =df.groupby("top_scorer")["highscore"].sum().sort_values(ascending=False).head(2)
high.plot( kind="bar",color="y")
```

3.10 Best Bowling Figures

```python
df.head(2)
```

```python
df["Highest_Wicket"] = df["best_bowling_figure"].apply(lambda x:x.split('--')[0])
df["Highest_Wicket"] = df["Highest_Wicket"].astype(int)
top_bowller = df.groupby("best_bowling")["Highest_Wicket"].sum().sort_values(ascending=False).head(10)
sns.barplot(y=top_bowller.index,x=top_bowller.values)
plt.title("Top 10 Bowller")
plt.xlabel("Wicket")
```
