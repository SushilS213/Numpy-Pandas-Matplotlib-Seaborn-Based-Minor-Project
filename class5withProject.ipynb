```python
df = pd.read_csv("anime.csv")
df.head()
```

```python
def extract_episodes(txt):
    check = False
    Data = ""
    for i in txt:
        if i==")":
          check = False
          return  Data
        if check ==True:
           Data=Data + i
        if i =="(":
           check = True        
```

```python
df["Episodes"] = df['Title'].apply(extract_episodes)
df["Episodes"] = df["Episodes"].str.replace(" eps","")
df.head()
```

```python
def time_Stamp(txt1):
    check = False
    data = ""
    for i in txt1:
        if i==",":
            check = False
            return data
        if check ==True:
            data = data+i
        if i == ")":
           check = True    

```

```python
df["Time_Stamp"]  = df["Title"].apply(time_Stamp)
df.head()
```

```python
def time_Stamp1(txt):
    check = False
    data = ""
    for i in range(len(txt)):
        if txt[i] == ")":
            for j in range(i+1,i+20):
                data +=txt[j]
            return data    
```

```python
df["Time_Stamp"]  = df["Title"].apply(time_Stamp1)
df.head()
```

```python

def calculate_date_diff(df, col_name):
    dates = df[col_name].str.split(' - ', expand=True)
    
    start = pd.to_datetime(dates[0], format='%b %Y')
    end = pd.to_datetime(dates[1], format='%b %Y')
    
    df['diff_days'] = (end - start).dt.days
    df['diff_months'] = (end.dt.year - start.dt.year) * 12 + (end.dt.month - start.dt.month)
    
    return df
```

```python


df1 = calculate_date_diff(df, 'Time_Stamp')
df1.head()
```
